```python
import pandas as pd

index = pd.read_csv('./data/icaird_cervical/index.csv')

groups = index.groupby('split')
splits = { name: group for name, group in groups }
```

```python
train_df = splits['train']
train_df.head() 
```

```python
index.groupby('subcategory').count()['category'].sort_values(ascending=False).plot.bar()
```

```python
NUM_SAMPLES_PER_CLASS = 14

sampled_groups = index.groupby('subcategory', group_keys=False).sample(n=NUM_SAMPLES_PER_CLASS, random_state=42)
sampled_groups
```

```python
sampled_groups.value_counts('subcategory')
```

```python
sampled = sampled_groups.sample(n=120, random_state=42)
sampled.value_counts('subcategory')
```

```python
sampled.to_csv('./data/icaird_cervical/sampled.csv', index=False)
```
